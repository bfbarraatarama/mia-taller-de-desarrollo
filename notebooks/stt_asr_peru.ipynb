# STT (Stage 3): `asr.json` de los ejemplos peruanos

Solo carga el `asr.json` de PER-1 y PER-2. Se generan con `python -m evaluacion.stt.ejemplos_peru`.

```python
import json
from src.rutas import ARCHIVO_ASR, DIRECTORIO_PROYECTO

OUTPUT_BASE = DIRECTORIO_PROYECTO / "evaluacion" / "stt" / "generado"

asr = {}
for test_id in ["PER-1", "PER-2"]:
    with open(OUTPUT_BASE / test_id / ARCHIVO_ASR, encoding="utf-8") as f:
        asr[test_id] = json.load(f)
```

```python
for test_id, data in asr.items():
    segmentos = data["segments"]
    n_palabras = sum(len(s["words"]) for s in segmentos)
    fines = [s["end"] for s in segmentos if s["end"] is not None]
    ultimo = f"{max(fines):.1f} s" if fines else "sin intervalo"
    sin_intervalo = sum(s["end"] is None for s in segmentos)
    print(f"{test_id}: idioma={data['language']}, {len(segmentos)} segmentos, {n_palabras} palabras, "
          f"hasta {ultimo}; {sin_intervalo} segmentos sin intervalo")
```

```python
for test_id, data in asr.items():
    print(f"--- {test_id} ---")
    for s in data["segments"][:10]:
        intervalo = (f"{s['start']:7.2f} - {s['end']:7.2f}"
                     if s["start"] is not None else "sin intervalo temporal")
        print(f"[{intervalo}] {s['text']}")
```

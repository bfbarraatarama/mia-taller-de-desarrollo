# STT (Stage 3): `asr.json` de los ejemplos peruanos

Solo carga el `asr.json` de PER-1 y PER-2. Se generan con `PYTHONPATH=src python -m STT.ejemplos_peru`.

```python
import json
import sys
from pathlib import Path

raiz = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "rutas.py").exists())
sys.path.insert(0, str(raiz / "src"))
from rutas import ARCHIVO_ASR, DIRECTORIO_PROYECTO

OUTPUT_BASE = DIRECTORIO_PROYECTO / "data" / "ASR" / "tmp" / "test_asr"

asr = {}
for test_id in ["PER-1", "PER-2"]:
    with open(OUTPUT_BASE / test_id / ARCHIVO_ASR, encoding="utf-8") as f:
        asr[test_id] = json.load(f)
```

```python
for test_id, data in asr.items():
    segmentos = data["segments"]
    n_palabras = sum(len(s["words"]) for s in segmentos)
    print(f"{test_id}: idioma={data['language']}, {len(segmentos)} segmentos, {n_palabras} palabras, "
          f"hasta {segmentos[-1]['end']:.1f} s")
```

```python
for test_id, data in asr.items():
    print(f"--- {test_id} ---")
    for s in data["segments"][:10]:
        print(f"[{s['start']:7.2f} - {s['end']:7.2f}] {s['text']}")
```
